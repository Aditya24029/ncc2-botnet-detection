# 04 - FT-Transformer | Sensor 3

Loads the Sensor 1 FT-Transformer and fine-tunes it on Sensor 3.

Split is 70/15/15: validation picks the checkpoint, test is used once.

```python
import sys
sys.path.append("..")

import pandas as pd

from src.common import (
    PROCESSED_DIR, MODELS_DIR, TARGET,
    CATEGORICAL_COLS, CONTINUOUS_COLS,
    three_way_split, report_and_save,
)

from pytorch_tabular import TabularModel
```

```python
df = pd.read_csv(f"{PROCESSED_DIR}/sensor3_processed.csv")
train_df, val_df, test_df = three_way_split(df)
print("train", train_df.shape, "| val", val_df.shape, "| test", test_df.shape)
```

```python
model = TabularModel.load_model(f"{MODELS_DIR}/ft_pretrained_sensor1")
model.fit(train=train_df, validation=val_df)   # validation = val, NOT test
model.save_model(f"{MODELS_DIR}/ft_finetuned_sensor3")
```

```python
pred = model.predict(test_df)
report_and_save("FT-Transformer", "sensor3", test_df[TARGET], pred[f"{TARGET}_prediction"])
```
